## 2. Tạo dữ liệu không gian đầu tiên

Ta tạo ba điểm mẫu theo hệ tọa độ WGS 84 (`EPSG:4326`) và đóng gói thành `GeoDataFrame`.

```python
import geopandas as gpd
from shapely.geometry import Point

dia_diem = gpd.GeoDataFrame(
    {"ten": ["Điểm A", "Điểm B", "Điểm C"]},
    geometry=[Point(106.34, 10.25), Point(106.40, 10.20), Point(106.28, 10.18)],
    crs="EPSG:4326",
)
dia_diem
```

## 3. Chuyển hệ tọa độ và tạo vùng đệm

Phép đo khoảng cách cần CRS phẳng. Ta chuyển sang UTM 48N (`EPSG:32648`) rồi tạo vùng đệm bán kính 2 km.

```python
dia_diem_utm = dia_diem.to_crs(32648)
vung_dem = dia_diem_utm.copy()
vung_dem.geometry = vung_dem.buffer(2_000)
vung_dem[["ten", "geometry"]]
```

## 4. Hiển thị bản đồ tương tác

Chuyển kết quả về WGS 84 và hiển thị trực tiếp trong notebook.

```python
ban_do = vung_dem.to_crs(4326).explore(
    column="ten",
    tooltip=["ten"],
    style_kwds={"fillOpacity": 0.25},
    legend=False,
)
dia_diem.explore(m=ban_do, color="#d9480f", marker_kwds={"radius": 7})
```
